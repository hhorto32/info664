## Bonus

```python
import csv
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) 
    data = list(reader)
print(data[1])
```

This was taken directly from the notes. For the work below, I made 4 different empty sets (this feels like a repetitive way to iterate, but it worked). Then, I ran a loop to fill each empty "set". I then counted the 4 sets that I made and created variables so that my f-string looked better. 

```python
female_artists = []
male_artists = []
non_binary_artists = []
other_artists = []
for artist in data:
    if artist["Gender"] == "female":
        female_artists.append(artist)
    elif artist["Gender"] == "male":
        male_artists.append(artist)
    elif artist["Gender"] == "non-binary":
        non_binary_artists.append(artist)
    else:
        other_artists.append(artist)
f_count = len(female_artists)
m_count = len(male_artists)
nb_count = len(non_binary_artists)
other_count = len(other_artists)
print(f_count)
print(m_count)
print(nb_count)
print(other_count)
```

```python
print(f"In the MoMa database, they had {f_count} female artists, {m_count} male artists, {nb_count} non-binary artists and {other_count} unspecified.")
```

Since I used variables for my length counts, I used those variables in my f-string.
