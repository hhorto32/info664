### ^ Searching whole database vs my keyword

```python
object_path = "/public/collection/v1/objects/"
horse_item = horse_object_ids[0]
url = f"{base_url}{object_path}{horse_item}"
print(url)
response_object = requests.get(url)
print(response_object.ok)
```

### Make a list with these ideas and put it all into 1 list.

```python
from time import sleep
object_path = "/public/collection/v1/objects/"

horse_details = []
for horse_id in horse_object_ids: # for each of the object IDs i pulled from the MET api earlier...
    url = f"{base_url}{object_path}{horse_id}"
    response_object = requests.get(url) # making the request to the server
    horse_details.append(response_object.json())
    print(f"grabbed details from object id {horse_id} with status code {response_object.status_code}")
    sleep(0.5)
```
