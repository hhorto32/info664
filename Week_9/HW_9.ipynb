## The code below checks how many contain 123 in the password. 
If the value is true, then it contains 123. If it is false, it doesn't contain it. 

```python
passcode  = df['password'].str.contains('123', na=False)
df[passcode]
df.value_counts(passcode)
```

```python
passcode1  = df['password'].str.endswith('a', na=False)
df.value_counts(passcode1)
```

## Predicting Next Character
#### Below shows how to have a predictor of the next character. From the password data frame, it first sorted all the ones that start with 250. Then, it looked for the next character (3), since we start at zero (so 250 would have spots: 0, 1,2). Then char_counts gives the predicted values as a bar chart. 

```python
alphapass1  = df['password'].str.startswith('250', na=False)
filtered_passwords = df[alphapass1].password

next_chars = filtered_passwords.str[3]

char_counts = next_chars.value_counts().plot(kind="barh")
char_counts
```
