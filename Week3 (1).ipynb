### I used 3 simple data collection points - money I spent on coffee, steps taken and my worldle score for one week from 9/6-9/13.


```python
money_coffee = [4.35, 14.20, 11.70, 3.50, 3.50, 6.53, 0.00] 
steps_taken = [13681, 12465, 14291, 8916, 6244, 6886, 19377, 11020]
wordle_score = [5,4,6, 3, 5, 5, 3, 6]
print (money_coffee)
print (steps_taken)
print(wordle_score)
```

## To sort it low to high, I used sort. 

```python
wordle_score.sort()
print(wordle_score)
```

## To sort it high to low, I used sort, reverse = true, telling the computer to reverse the order. 

```python
steps_taken.sort(reverse = True)
print(steps_taken)
```

## Added a new data entry

```python
new_coffee = 4.74
print(new_coffee)
```

```python
money_coffee.append(new_coffee)
print(money_coffee)
```

## I Googled "extend", because I wanted to see how I could merge two lists. It appears that extend allowed me to concatenate (but with appropriate spacing) the two lists.

```python
new_coffee_list = [4.74, 1.47, 4.74]
money_coffee.extend(new_coffee_list)
print(money_coffee)
```
